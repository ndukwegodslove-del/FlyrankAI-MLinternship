## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a ranking problem because the goal is to prioritize pages for content action, rather than simply assign every page to a class. A simple rule can use observable features such as impressions and days since last update to give pages different scores and rank them. With enough labelled data, a supervised ML model can learn patterns from the available features and produce scores that can also be used to rank pages. I can evaluate whether the ranking puts more declining pages near the top using Precision@K.

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/ndukwegodslove-del/FlyrankAI-MLinternship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_URL)

print("Rows:", len(df))
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target will be whether a page is declining, represented by is_declining_label. A value of 1 means the page's trend_direction is "down", while 0 means it is not. This serves as a proxy for identifying pages that may need content attention. The target is derived from the observed trend_direction rather than from a direct measurement of whether a content refresh was successful.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use Precision@50 to evaluate the ranking. It measures the proportion of the top 50 pages ranked by the model that are actually labelled as declining. A higher Precision@50 means that more of the pages prioritized for review are genuinely declining, making the ranking more useful for directing limited content-review resources.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
df[[
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count",
    "trend_pct"
]].head()
```

The unit of analysis is one content page. Each row in the dataset represents a page and contains observable features describing that page, such as impressions, content age, days since last update, average position, CTR and word count. The target indicates whether the page is declining. Therefore, the ML model will produce a score for each page, which can then be used to rank the pages for content attention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as “if a page has at least 500 impressions in the last 90 days and has not been updated for 180 days or more, classify it as high priority” is too rigid. Small differences around the thresholds can cause pages with very similar characteristics to receive different classifications. The decision may also depend on several variables at the same time, rather than just two fixed conditions. ML could potentially learn these patterns from historical data instead of relying on manually selected cutoffs.
